Question 1: What is Anomaly Detection? Explain its types (point, contextual, and collective anomalies) with examples.

Ans:- Anomaly Detection is the process of identifying rare, unusual, or abnormal data points that significantly differ from the normal pattern of a dataset.

These unusual observations are called anomalies or outliers, and they often indicate:

- Fraudulent activity
- System failures
- Network intrusions
- Medical abnormalities
- Data errors

Anomalies are generally classified into three main types:

1. Point Anomalies
- Definition:

A single data point that is significantly different from the rest of the data.

example:- in a dataset of daily transacton
| Day | Amount ($) |
| --- | ---------- |
| 1   | 200        |
| 2   | 180        |
| 3   | 220        |
| 4   | 4000       |


Transaction of 4000 is a point anomaly

Real-world Use Cases:
- Credit card fraud detection
- Sensor malfunction (sudden spike)
- Unusual temperature reading

2. Contextual Anomalies (Conditional Anomalies)
- Definition:

A data point is considered anomalous only in a specific context, such as time, season, or location.

Example:
- Temperature reading:

| Season | Temperature |
| ------ | ----------- |
| Summer | 40°C        |
| Winter | 40°C        |

- 40°C in summer is normal, but
- 40°C in winter is abnormal → Contextual anomaly.

Context Factors:
- Time (day/night, weekday/weekend)
- Location
- Weather/season

Real-world Use Cases:

- Sudden high electricity usage at midnight

- Unusual website traffic at odd hours

- Medical heart rate changes during rest

3. Collective Anomalies
- Definition:

A group of data points together forms an anomaly, even if individual points may look normal.

Example:

In ECG heart monitoring:
A single heartbeat may look normal, but a sequence of irregular beats indicates an anomaly (arrhythmia).

Real-world Use Cases:
- Detecting cyber-attacks (DDoS patterns)
- Faults in industrial machines over time
- Abnormal customer behavior sequences




Question 2: Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of their approach and suitable use cases.

Ans:- Comparison: Isolation Forest vs DBSCAN vs Local Outlier Factor

1. Isolation Forest (IF)
- Approach

Isolation Forest is based on the idea that:

Anomalies are easier to isolate than normal points.

- It builds many random decision trees.
- Outliers require fewer splits to be separated.
- The shorter the path length, the more anomalous the point.

a. Strengths
- Works well in high-dimensional data
- Efficient for large datasets
- Does not rely on distance calculations
- Handles global outliers effectively

b. Weaknesses
- Less effective when anomalies are not clearly separable
- Does not capture local density variations well

c. Suitable Use Cases
- Fraud detection in banking
- Network intrusion detection
- Large-scale monitoring systems
- High-dimensional datasets (many features)

2. DBSCAN (Density-Based Clustering)
- Approach

DBSCAN detects anomalies by clustering:

Points in dense regions form clusters, and points in sparse regions are outliers.

- Requires two parameters:
- eps (neighborhood radius)
- min_samples (minimum points to form a dense region)
- Points not belonging to any cluster are labeled as anomalies.

a. Strengths
- Finds clusters of arbitrary shapes
- Detects noise naturally (outliers)
- No need to specify number of clusters

b. Weaknesses
- Sensitive to parameter tuning
- Struggles with varying density datasets
- Poor performance in high dimensions

c. Suitable Use Cases
- Spatial anomaly detection (GPS, maps)
- Image segmentation noise detection
- Outlier detection in low-dimensional data

3. Local Outlier Factor (LOF)
- Approach

LOF is based on local density:

A point is an outlier if its density is much lower than its neighbors.

- Compares density of each point with surrounding points.
- Detects anomalies relative to local neighborhoods.

a. Strengths
- Excellent for detecting local anomalies
- Works well when data has varying density regions

b. Weaknesses
- Computationally expensive for large datasets
- Not ideal for very high-dimensional data
- Sensitive to choice of neighbors (k)

c. Suitable Use Cases
- Detecting abnormal customer behavior in segments
- Sensor anomalies in local environments
- Healthcare monitoring (patient-specific deviations)

Question 3: What are the key components of a Time Series? Explain each with one example.

Ans:- Key Components of a Time Series

A Time Series is a sequence of data points recorded at regular time intervals (daily, monthly, yearly, etc.).

Example:
- Monthly sales, stock prices, temperature readings, website traffic.

A time series is usually made up of four main components:
1. Trend(T)


The long-term movement of the data over time — upward or downward direction.

Example:
- Company sales increasing steadily every year:
| Year | Sales |
| ---- | ----- |
| 2020 | 50K   |
| 2021 | 65K   |
| 2022 | 80K   |
This upward growth shows a trend.
2. Seasonality (S)

Regular and repeating patterns that occur at fixed intervals (daily, monthly, yearly)

Example:
- Ice cream sales peak every summer:
| Month | Sales |
| ----- | ----- |
| June  | High  |
| July  | High  |
| Dec   | Low   |
This repeating yearly cycle is seasonality.
3. Cyclic component (C)

Fluctuations that occur over longer periods, but not fixed or regular like seasonality.

Example:
- Economic cycles:
- Boom → Recession → Recovery → Boom
- Stock market rises and falls over several years.
- These multi-year patterns are cyclical.

4. Irragular / Noise (I)

Meaning:
- Random, unpredictable variations caused by unexpected events.
Example:
- Sudden drop in airline bookings due to a pandemic.
- Unexpected spike in demand due to viral marketing.
- Such sudden changes are irregular components.



Question 4: Define Stationary in time series. How can you test and transform a non-stationary series into a stationary one?

Ans:- Stationary Time Series

A stationary time series is a series whose statistical properties remain constant over time.

In a stationary series:

- Mean is constant
- Variance is constant
- Autocovariance structure does not change with time

In simple words:

A stationary series has no long-term trend or changing seasonality, and its behavior stays stable over time.

How to Test Stationarity?
1. Visual Inspection
Check:

- Trend
- Seasonality
- Changing variance

Plot the time series:
- Upward slope → non-stationary
- Seasonal cycles → non-stationary

2. Rolling Statistics

Compute:
- Rolling mean
- Rolling variance

If they change over time → non-stationary.

3. ADF Test (Augmented Dickey-Fuller Test)

Most common statistical test.

Hypotheses:
- H₀ (Null): Series is non-stationary
- H₁ (Alternative): Series is stationary

Interpretation:
- If p-value < 0.05 → Reject H₀ → Stationary
- If p-value ≥ 0.05 → Non-stationary

4. KPSS Test

Opposite of ADF:
- H₀: Series is stationary
- H₁: Series is non-stationary

Used along with ADF for confirmation.

How to Transform a Non-Stationary Series into Stationary?
1. Differencing

Removes trend by subtracting previous value:

Yt′​=Yt​−Yt−1​

Original: 100, 150, 200
Differenced: 50, 50

2. Seasonal Differencing

Removes seasonal patterns:

Yt′​=Yt​−Yt−s​

Where s = seasonal period (e.g., 12 months)

3. Log Transformation

Used when variance increases over time.

Yt′​=log(Yt​)

Example:
- Stock prices often need log scaling.



Question 5: Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in terms of structure and application.

Ans:- Differentiation Between AR, MA, ARIMA, SARIMA, SARIMAX

Time series forecasting models are mainly built using three ideas:

- Autoregression (AR)
- Moving Average (MA)
- Differencing (I)
- Seasonality (S)
- External variables (X)

1. AR Model (Autoregressive)

Structure
- AR model predicts future values using past values of the series.

Yt​=c+ϕ1​Yt−1​+ϕ2​Yt−2​+...+ϕp​Yt−p​+ϵt​

p = number of lag observations

Application

- Stock prices (short-term dependence)
- Temperature forecasting

2. MA Model (Moving Average)

Structure
- MA model predicts using past forecast errors.

Yt​=c+θ1​ϵt−1​+θ2​ϵt−2​+...+θq​ϵt−q​

- q = number of error lags

Application

- Noise smoothing
- Modeling sudden random effects

3. ARIMA Model (Autoregressive Integrated Moving Average)

Structure

ARIMA combines:
- AR (p)
- Differencing (d)
- MA (q)

ARIMA(p,d,q)

Application
- Sales forecasting with trend
- Economic indicators forecasting

4. SARIMA Model (Seasonal ARIMA)

Structure
- SARIMA extends ARIMA by adding seasonal terms.

SARIMA(p,d,q)(P,D,Q,s)

Where:
- Seasonal AR = P
- Seasonal differencing = D
- Seasonal MA = Q
- s = seasonal period (e.g., 12 for monthly data)

Application

- Monthly retail sales
- Electricity demand (daily/annual cycles)

5. SARIMAX Model (Seasonal ARIMA with Exogenous Variables)

Structure

- SARIMAX = SARIMA + External predictors (X)

SARIMAX(p,d,q)(P,D,Q,s)+X

Application
- Sales prediction using advertising spend
- Weather forecasting using humidity, wind, etc.
- Demand forecasting with holidays/events

```python
# Question 8: Train a SARIMA model on the monthly airline passengers dataset. Forecast the next 12 months and visualize the results.
```

```python
import pandas as pd

# Load the AirPassengers dataset from the local file
df = pd.read_csv('/content/AirPassengers.csv', parse_dates=['Month'])

# Set the 'Month' column as the DataFrame's index
df.set_index('Month', inplace=True)
df.index = pd.to_datetime(df.index) # Ensure index is datetime type

# Rename the '#Passengers' column to 'Passengers' for clarity
df.rename(columns={'#Passengers': 'Passengers'}, inplace=True)

print("AirPassengers dataset loaded successfully from /content/AirPassengers.csv.")
print("\nDataFrame Head:")
print(df.head())
print("\nDataFrame Info:")
df.info()

import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

# Function to perform ADF test and print results
def adf_test(series, title='Augmented Dickey-Fuler Test'):
    print(f"--- {title} ---")
    result = adfuller(series.dropna())
    labels = ['ADF Statistic', 'p-value', '#Lags Used', 'Number of Observations Used']
    for value, label in zip(result, labels):
        print(f'{label}: {value}')
    if result[1] <= 0.05:
        print("Strong evidence against the null hypothesis, series is stationary.")
    else:
        print("Weak evidence against the null hypothesis, series is non-stationary.")
    print("\n")

# Original Series
print("Original Series:")
adf_test(df['Passengers'])

# Plot ACF and PACF for the original series
fig, axes = plt.subplots(2, 1, figsize=(12, 8))
plot_acf(df['Passengers'], ax=axes[0], lags=40)
plot_pacf(df['Passengers'], ax=axes[1], lags=40)
axes[0].set_title('ACF of Original Series')
axes[1].set_title('PACF of Original Series')
plt.tight_layout()
plt.show()

# Apply first-order differencing
df['Passengers_diff1'] = df['Passengers'].diff().dropna()

print("First-order Differenced Series:")
adf_test(df['Passengers_diff1'], title='ADF Test on First-order Differenced Series')

# Plot ACF and PACF for the first-order differenced series
fig, axes = plt.subplots(2, 1, figsize=(12, 8))
plot_acf(df['Passengers_diff1'], ax=axes[0], lags=40)
plot_pacf(df['Passengers_diff1'], ax=axes[1], lags=40)
axes[0].set_title('ACF of First-order Differenced Series')
axes[1].set_title('PACF of First-order Differenced Series')
plt.tight_layout()
plt.show()

# Apply seasonal differencing (period=12)
df['Passengers_diff1_seasonal12'] = df['Passengers_diff1'].diff(12).dropna()

print("First-order and Seasonal Differenced Series:")
adf_test(df['Passengers_diff1_seasonal12'], title='ADF Test on First-order and Seasonal Differenced Series')

# Plot ACF and PACF for the first-order and seasonal differenced series
fig, axes = plt.subplots(2, 1, figsize=(12, 8))
plot_acf(df['Passengers_diff1_seasonal12'], ax=axes[0], lags=40)
plot_pacf(df['Passengers_diff1_seasonal12'], ax=axes[1], lags=40)
axes[0].set_title('ACF of First-order and Seasonal Differenced Series')
axes[1].set_title('PACF of First-order and Seasonal Differenced Series')
plt.tight_layout()
plt.show()

from statsmodels.tsa.statespace.sarimax import SARIMAX

# Instantiate the SARIMAX model
# Based on the ACF/PACF plots and ADF tests, suitable orders are (1,1,1) for non-seasonal
# and (1,1,1,12) for seasonal components.
# d=1 from first differencing, D=1 from seasonal differencing.
# p and q are chosen by examining the ACF/PACF plots after differencing.
model = SARIMAX(df['Passengers'],
                order=(1, 1, 1),
                seasonal_order=(1, 1, 1, 12),
                enforce_stationarity=False,
                enforce_invertibility=False)

# Fit the SARIMAX model
model_fit = model.fit()

# Print the model summary
print(model_fit.summary())

print("SARIMA model trained and summary displayed.")

import pandas as pd
import matplotlib.pyplot as plt

# Generate forecasts for the next 12 months
# `steps` specifies the number of periods to forecast
forecast_steps = 12
forecast = model_fit.get_forecast(steps=forecast_steps)

# Get confidence intervals for the forecast
forecast_ci = forecast.conf_int()

# Get the forecasted values
forecast_values = forecast.predicted_mean

# Print the forecasted values and confidence intervals
print("Forecasted values for the next 12 months:")
print(forecast_values)
print("\n95% Confidence Intervals:")
print(forecast_ci)

# Plot the original series, forecast, and confidence intervals
plt.figure(figsize=(15, 7))
plt.plot(df['Passengers'], label='Original Series')
plt.plot(forecast_values.index, forecast_values, label='Forecast', color='red')
plt.fill_between(forecast_ci.index, forecast_ci.iloc[:, 0], forecast_ci.iloc[:, 1], color='pink', alpha=0.3, label='95% Confidence Interval')

plt.title(f'SARIMA Forecast of Air Passengers for Next {forecast_steps} Months')
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.legend()
plt.grid(True)
plt.show()

print("SARIMA forecast generated and visualized.")
```

Energy demand depends on:
- Past usage
- Daily/weekly seasonality
- Weather conditions
- Region-wise behavior

**Model Options**

**ARIMA (Trend-Based Forecasting)**

Works for:
- Single series without seasonality
- Stationary or differenced data

Limitation:
- Cannot handle strong seasonal demand patterns

**SARIMA (Seasonal ARIMA)**

Works for:
- Energy demand with daily/weekly cycles

SARIMA(p,d,q)(P,D,Q,s)

Use Case:
- Forecasting demand with repeating patterns (peak hours)

**SARIMAX (Best Choice for Power Grid Forecasting)**

Why SARIMAX?

SARIMAX includes:

- Trend + seasonality
- External variables (weather, holidays, region)

SARIMAX=SARIMA+X

Example Features (X):
- Temperature
- Humidity
- Weekend indicator
- Region demand profile

Use Case:
- Demand increases due to extreme heat → more AC usage.

**Validation and Performance Monitoring Over Time**

Forecast Validation

Metrics:

- MAE (Mean Absolute Error)

- RMSE (Root Mean Squared Error)

- MAPE (% error)

Approach:

- Rolling-window cross-validation
- Train on past → test on next few hours/days

Anomaly Detection Validation

- Use historical labeled anomalies (if available)
- Precision / Recall / F1-score
- Monitor false alarms

Model Drift Monitoring

Energy consumption patterns change due to:
- New industries
- Seasonal shifts
- Policy changes

So we monitor:
- Error increase over time
- Anomaly frequency spikes
- Distribution shift

Real-Time Monitoring Dashboard

A dashboard for grid operators showing:
- Forecast vs actual demand
- Anomaly alerts
- Region-wise consumption spikes
- Model confidence intervals

Retraining Strategy
- Weekly retraining for forecasting models
- Online updates for anomaly thresholds

**Business and Operational Benefits**

Operational Benefits
- Early warning for equipment failures
- Prevent overload and blackouts
- Real-time outage detection

Business Benefits
- Better energy production planning
- Reduce operational cost by balancing supply-demand
- Detect energy theft or abnormal usage

Decision Support

Forecast helps:
- Schedule power generation
- Optimize renewable integration (solar/wind)
- Plan peak-hour pricing



```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler

from statsmodels.tsa.statespace.sarimax import SARIMAX
from sklearn.metrics import mean_absolute_error, mean_squared_error

df = pd.read_csv("/content/energy_data (1).csv")

# Print columns to debug KeyError: 'timestamp'
print("DataFrame columns after loading CSV:", df.columns)

# Rename columns based on assumed mapping (T: temperature, H: humidity, PO: energy_usage)
df.rename(columns={'T': 'temperature', 'H': 'humidity', 'PO': 'energy_usage'}, inplace=True)

# Create a DatetimeIndex assuming 15-minute intervals. We'll start from a arbitrary date.
start_date = '2023-01-01 00:00:00'
df['timestamp'] = pd.date_range(start=start_date, periods=len(df), freq='15min')
df.set_index('timestamp', inplace=True)

print(df.head())

# Add time-based features
df["hour"] = df.index.hour
df["dayofweek"] = df.index.dayofweek
df["month"] = df.index.month

# Fill missing values (if any after column renaming)
df = df.interpolate()

print(df.isnull().sum())
features = ["energy_usage", "temperature", "humidity"]

X = df[features]

# Scale data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Isolation Forest Model
iso_forest = IsolationForest(
    n_estimators=200,
    contamination=0.02,
    random_state=42
)

df["anomaly_iforest"] = iso_forest.fit_predict(X_scaled)

# -1 means anomaly
df["anomaly_iforest"] = df["anomaly_iforest"].map({1: 0, -1: 1})

print(df["anomaly_iforest"].value_counts())

plt.figure(figsize=(12,5))
plt.plot(df.index, df["energy_usage"], label="Energy Usage")

# Highlight anomalies
plt.scatter(
    df[df["anomaly_iforest"] == 1].index,
    df[df["anomaly_iforest"] == 1]["energy_usage"],
    label="Anomalies",
    marker="o"
)

plt.title("Isolation Forest Anomaly Detection")
plt.legend()
plt.show()

lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.02
)

df["anomaly_lof"] = lof.fit_predict(X_scaled)
df["anomaly_lof"] = df["anomaly_lof"].map({1: 0, -1: 1})

print(df["anomaly_lof"].value_counts())

dbscan = DBSCAN(eps=0.8, min_samples=10)

df["cluster_dbscan"] = dbscan.fit_predict(X_scaled)

# Points labeled -1 are anomalies
df["anomaly_dbscan"] = df["cluster_dbscan"].apply(lambda x: 1 if x == -1 else 0)

print(df["anomaly_dbscan"].value_counts())
train_size = int(len(df) * 0.8)

train = df.iloc[:train_size]
test  = df.iloc[train_size:]

y_train = train["energy_usage"]
y_test  = test["energy_usage"]

# Exogenous variables (weather and time-based features)
exog_train = train[["temperature", "humidity", "hour", "dayofweek", "month"]]
exog_test  = test[["temperature", "humidity", "hour", "dayofweek", "month"]]

model = SARIMAX(
    y_train,
    exog=exog_train,
    order=(2,1,2),          # ARIMA(p,d,q)
    seasonal_order=(1,1,1,96)  # 96 intervals/day (15 min data)
)

sarimax_result = model.fit()

print(sarimax_result.summary())

forecast = sarimax_result.forecast(
    steps=len(test),
    exog=exog_test
)

test["forecast"] = forecast

plt.figure(figsize=(12,5))

plt.plot(test.index, y_test, label="Actual Demand")
plt.plot(test.index, test["forecast"], label="Forecast Demand")

plt.title("SARIMAX Forecasting (15-min Demand)")
plt.legend()
plt.show()
mae = mean_absolute_error(y_test, forecast)
rmse = np.sqrt(mean_squared_error(y_test, forecast))

print("Forecast Performance:")
print("MAE :", mae)
print("RMSE:", rmse)

def monitor_new_data(new_point):
    """
    new_point: dictionary containing new usage + weather values
    """

    # Convert to DataFrame
    new_df = pd.DataFrame([new_point])

    # Scale for anomaly detection
    new_scaled = scaler.transform(new_df[features])

    # Predict anomaly
    anomaly = iso_forest.predict(new_scaled)

    if anomaly == -1:
        print("⚠ ALERT: Abnormal Energy Spike/Drop Detected!")

    else:
        print("✅ Normal Consumption Pattern")

    return anomaly

new_data = {
    "energy_usage": 550,
    "temperature": 42,
    "humidity": 80
}

monitor_new_data(new_data)
```
